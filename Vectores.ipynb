```python
import numpy as np

# Crear un vector
vector = np.array([3, 4])

print("Mi vector es:", vector)
print("Primera componente:", vector[0])
print("Segunda componente:", vector[1])
print("Número de componentes:", len(vector))
```

```python
vector_3d = np.array([2, 5, 7])

print("Vector:", vector_3d)
print("Componente x:", vector_3d[0])
print("Componente y:", vector_3d[1])
print("Componente z:", vector_3d[2])
```

```python
import numpy as np

a = np.array([2, 3])
b = np.array([4, 1])

suma = a + b

print("Vector a:", a)
print("Vector b:", b)
print("Suma:", suma)
```

```python
import numpy as np

a = np.array([8, 6])
b = np.array([3, 2])

resta = a - b

print("Vector a:", a)
print("Vector b:", b)
print("Resta:", resta)
```

```python
import numpy as np

vector = np.array([2, 3])
escalar = 3

resultado = escalar * vector

print("Vector original:", vector)
print("Escalar:", escalar)
print("Resultado:", resultado)
```

```python
import numpy as np

v = np.array([3, -2])

opuesto = -v
resultado = v + opuesto

print("Vector original:", v)
print("Vector opuesto:", opuesto)
print("Suma:", resultado)
```

```python
import numpy as np

v = np.array([3, 4])

magnitud = np.linalg.norm(v)

if magnitud != 0:
    unitario = v / magnitud
    print("Vector original:", v)
    print("Vector unitario:", unitario)
    print("Magnitud del unitario:", np.linalg.norm(unitario))
else:
    print("El vector cero no tiene un vector unitario definido.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Componentes del vector
x = 3
y = 4

plt.figure(figsize=(7, 6))

# Dibujar la flecha desde el origen
plt.quiver(
    0, 0, x, y,
    angles="xy",
    scale_units="xy",
    scale=1,
    color="blue"
)

# Configuración de los ejes
plt.xlim(-1, 5)
plt.ylim(-1, 6)
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)
plt.grid(True)

plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.title("Representación del vector (3, 4)")
plt.gca().set_aspect("equal", adjustable="box")

plt.show()
```

```python
import matplotlib.pyplot as plt

vectores = [
    (2, 3, "red"),
    (4, 1, "green"),
    (-2, 2, "purple")
]

plt.figure(figsize=(7, 6))

for x, y, color in vectores:
    plt.quiver(
        0, 0, x, y,
        angles="xy",
        scale_units="xy",
        scale=1,
        color=color
    )

plt.xlim(-4, 6)
plt.ylim(-2, 5)
plt.axhline(0, color="black")
plt.axvline(0, color="black")
plt.grid(True)
plt.gca().set_aspect("equal", adjustable="box")
plt.title("Comparación de varios vectores")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

# Vector desde el origen hasta (2, 3, 4)
ax.quiver(0, 0, 0, 2, 3, 4, color="red", arrow_length_ratio=0.12)

ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_zlim(0, 5)

ax.set_xlabel("Eje X")
ax.set_ylabel("Eje Y")
ax.set_zlabel("Eje Z")
ax.set_title("Vector tridimensional (2, 3, 4)")

plt.show()
```
