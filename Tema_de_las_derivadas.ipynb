```python
import numpy as np
import matplotlib.pyplot as plt

# Crear valores de x
x = np.linspace(-5, 5, 200)

# Definir la función
y = x**2

# Graficar
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r"$f(x)=x^2$")

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Función f(x) = x²")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Función
def f(x):
    return x**2

# Punto inicial
a = 2

# Distancia entre los puntos
h = 2

# Segundo punto
x2 = a + h

# Coordenadas
y1 = f(a)
y2 = f(x2)

# Pendiente de la secante
m = (y2 - y1) / (x2 - a)

print("Punto P:", (a, y1))
print("Punto Q:", (x2, y2))
print("Pendiente de la secante:", m)

# Valores para la curva
x = np.linspace(0, 5, 200)
y = f(x)

# Valores para la recta secante
x_sec = np.linspace(0, 5, 100)
y_sec = y1 + m * (x_sec - a)

# Gráfica
plt.figure(figsize=(9, 6))

plt.plot(x, y, label="f(x) = x²")
plt.plot(x_sec, y_sec, "--", label="Recta secante")

plt.scatter([a, x2], [y1, y2], s=80)

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Recta secante")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import sympy as sp

# Crear variable
x = sp.symbols('x')

# Definir función
f = x**2

# Calcular derivada
derivada = sp.diff(f, x)

print("Función:")
sp.pprint(f)

print("\nDerivada:")
sp.pprint(derivada)
```

```python
import sympy as sp

x, h = sp.symbols('x h')

f = x**3

# Aplicamos directamente la definición
expresion = ((x + h)**3 - x**3) / h

print("Expresión de la definición:")
sp.pprint(sp.expand(expresion))

# Calcular el límite
derivada = sp.limit(expresion, h, 0)

print("\nDerivada:")
sp.pprint(derivada)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

def df(x):
    return 2*x

# Punto donde queremos la tangente
a = 2

# Punto de la curva
y_a = f(a)

# Pendiente de la tangente
m = df(a)

print("x =", a)
print("f(x) =", y_a)
print("f'(x) =", m)

# Ecuación de la tangente:
# y - f(a) = f'(a)(x-a)

x = np.linspace(0, 4, 200)

y = f(x)

y_tangente = y_a + m * (x - a)

plt.figure(figsize=(9, 6))

plt.plot(x, y, label="f(x) = x²")
plt.plot(x, y_tangente, "--", label="Recta tangente")

plt.scatter(a, y_a, s=100)

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Recta tangente a f(x)=x² en x=2")

plt.grid(True)
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Valores de x
x = np.linspace(-3, 3, 400)

# Función
f = x**3 - 3*x

# Derivada
df = 3*x**2 - 3

plt.figure(figsize=(10, 6))

plt.plot(x, f, label="f(x) = x³ - 3x")
plt.plot(x, df, label="f'(x) = 3x² - 3")

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Función y su derivada")

plt.grid(True)
plt.legend()

plt.show()
```

```python
import sympy as sp

x = sp.symbols('x')

f = x**2 - 4*x + 3

df = sp.diff(f, x)

puntos_criticos = sp.solve(df, x)

print("Función:")
sp.pprint(f)

print("\nDerivada:")
sp.pprint(df)

print("\nPuntos críticos:")
print(puntos_criticos)
```

```python
import sympy as sp

x = sp.symbols('x')

f = x**3 - 3*x

primera = sp.diff(f, x)
segunda = sp.diff(primera, x)

print("f(x) =")
sp.pprint(f)

print("\nf'(x) =")
sp.pprint(primera)

print("\nf''(x) =")
sp.pprint(segunda)
```

```python
import sympy as sp

x = sp.symbols('x')

f = x**3 - 3*x

# Derivadas
f1 = sp.diff(f, x)
f2 = sp.diff(f1, x)

# Puntos críticos
criticos = sp.solve(f1, x)

print("Función:")
sp.pprint(f)

print("\nPrimera derivada:")
sp.pprint(f1)

print("\nSegunda derivada:")
sp.pprint(f2)

print("\nPuntos críticos:")

for punto in criticos:
    valor_y = f.subs(x, punto)
    valor_segunda = f2.subs(x, punto)

    print("\nx =", punto)
    print("y =", valor_y)
    print("f''(x) =", valor_segunda)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

a = 2

valores_h = [1, 0.5, 0.1, 0.01]

for h in valores_h:

    pendiente = (f(a + h) - f(a)) / h

    print(f"h = {h}")
    print(f"Pendiente = {pendiente}")
    print()

# Gráfica
x = np.linspace(0, 4, 200)

plt.figure(figsize=(10, 6))

plt.plot(x, f(x), label="f(x) = x²")

for h in valores_h:

    x2 = a + h

    pendiente = (f(x2) - f(a)) / h

    y_secante = f(a) + pendiente * (x - a)

    plt.plot(x, y_secante, "--", label=f"h = {h}")

plt.scatter(a, f(a), s=100)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Aproximación de la recta tangente")

plt.grid(True)
plt.legend()

plt.show()
```

# 📚 Fórmulas principales

## Definición de derivada

f'(x) = lim(h→0) [f(x+h)-f(x)] / h

## Regla de la potencia

d/dx [xⁿ] = n xⁿ⁻¹

## Constante

d/dx [c] = 0

## Constante multiplicando

d/dx [c f(x)] = c f'(x)

## Suma

d/dx [f(x)+g(x)] = f'(x)+g'(x)

## Producto

d/dx [f(x)g(x)] = f'(x)g(x)+f(x)g'(x)

## Cociente

d/dx [f(x)/g(x)] =
[f'(x)g(x)-f(x)g'(x)] / g(x)²

## Regla de la cadena

d/dx [f(g(x))] =
f'(g(x))g'(x)

## Trigonométricas

d/dx [sin(x)] = cos(x)

d/dx [cos(x)] = -sin(x)

d/dx [tan(x)] = sec²(x)

# 🎯 Conclusión

La derivada es una herramienta fundamental del cálculo diferencial.

Su definición se obtiene mediante el límite:

f'(x) = lim(h→0) [f(x+h)-f(x)] / h

Geométricamente, la derivada representa la pendiente de la recta
tangente a una función en un punto.

Además, la derivada puede interpretarse como una razón de cambio
instantánea.

Por ejemplo:

- En física puede representar la velocidad.
- La segunda derivada puede representar la aceleración.
- En economía puede representar cambios marginales.
- En ingeniería permite analizar sistemas y cambios.
- En inteligencia artificial es fundamental para optimización y
  entrenamiento de modelos.

La idea central que debemos recordar es:

       DERIVADA = RAZÓN DE CAMBIO INSTANTÁNEA

y geométricamente:

       DERIVADA = PENDIENTE DE LA RECTA TANGENTE
