```python
import numpy as np
import matplotlib.pyplot as plt

# Funcion original
def f(x):
    return x**2 - 10*x + 16

# Derivada de la funcion
def derivada(x):
    return 2*x - 10

# Tasa de aprendizaje
beta = 0.2

# Funcion del gradiente descendente
def gradiente_descendente(x_inicial):
    x = x_inicial
    puntos = [x]

    print("\nValor inicial:", x)

    for i in range(1, 5):
        x_nuevo = x - beta * derivada(x)
        x = x_nuevo
        puntos.append(x)

        print("Iteracion", i,
              "| X =", round(x, 6),
              "| f(X) =", round(f(x), 6))

    return puntos

# Caso 1: iniciar en X = 0
puntos_0 = gradiente_descendente(0)

# Caso 2: iniciar en X = 8
puntos_8 = gradiente_descendente(8)

# Graficar la funcion
x = np.linspace(-1, 11, 400)

plt.figure(figsize=(10, 6))
plt.plot(x, f(x), label="f(x) = x^2 - 10x + 16")

# Graficar los puntos del caso 1
plt.plot(puntos_0, [f(v) for v in puntos_0],
         'o-', label="Inicio X = 0")

# Graficar los puntos del caso 2
plt.plot(puntos_8, [f(v) for v in puntos_8],
         's-', label="Inicio X = 8")

# Marcar el minimo
plt.scatter(5, -9, color='red',
            s=100, label="Minimo (5, -9)")

plt.xlabel("X")
plt.ylabel("f(X)")
plt.title("Algoritmo de Gradiente Descendente")
plt.grid()
plt.legend()
plt.show()
```
