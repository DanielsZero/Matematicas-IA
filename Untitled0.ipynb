Introducción a vectores y matrices

¿Qué es un vector?

Un vector es una estructura que permite almacenar varios valores en una sola variable.
En matemáticas, un vector es un objeto que puede utilizarse para representar una cantidad que tiene diferentes componentes.

En un nivel introductorio podemos pensar en un vector como una secuencia ordenada de números.

Por ejemplo:

$$ v = [3,5,2] $$

Este vector tiene 3 componentes:

Primera componente: 3
Segunda componente: 5
Tercera componente: 2
En Python podemos representarlo mediante un arreglo de NumPy.

Por ejemplo, un vector de tres elementos:

```python
vector = np.array([2, 4, 6])

print("Vector:", vector)
```

Suma de vectores usando matrices

Dos vectores se pueden sumar si tienen la misma cantidad de componentes. La suma se realiza componente por componente.

Por ejemplo, tenemos los vectores:

$$ A= \begin{bmatrix} 2\\ 4\\ 6 \end{bmatrix} $$

y

$$ B= \begin{bmatrix} 1\\ 3\\ 5 \end{bmatrix} $$

Para sumarlos, sumamos cada componente correspondiente:

$$ A+B= \begin{bmatrix} 2\\ 4\\ 6 \end{bmatrix} + \begin{bmatrix} 1\\ 3\\ 5 \end{bmatrix} $$ $$ A+B= \begin{bmatrix} 2+1\\ 4+3\\ 6+5 \end{bmatrix} $$

Por lo tanto:

$$ \boxed{ A+B= \begin{bmatrix} 3\\ 7\\ 11 \end{bmatrix} } $$
Ejemplo en Python

```python
import numpy as np

A = np.array([
    [2],
    [4],
    [6]
])

B = np.array([
    [1],
    [3],
    [5]
])

C = A + B

print("Vector A:")
print(A)

print("\nVector B:")
print(B)

print("\nA + B:")
print(C)
```

Suma de vectores representados como filas

También podemos representar los vectores horizontalmente:

$$ A= \begin{bmatrix} 2&4&6 \end{bmatrix} $$ $$ B= \begin{bmatrix} 1&3&5 \end{bmatrix} $$

Entonces:

$$ A+B= \begin{bmatrix} 2+1&4+3&6+5 \end{bmatrix} $$ $$ A+B= \begin{bmatrix} 3&7&11 \end{bmatrix} $$

En Python:

```python
A = np.array([[2, 4, 6]])
B = np.array([[1, 3, 5]])

C = A + B

print("A:")
print(A)

print("\nB:")
print(B)

print("\nA + B:")
print(C)
```

Ejemplo con matrices de 2 × 2

Aquí podemos aprovechar para conectar el concepto de vector con el de matriz.

Tenemos:

$$ A= \begin{bmatrix} 1&2\\ 3&4 \end{bmatrix} $$

y:

$$ B= \begin{bmatrix} 5&6\\ 7&8 \end{bmatrix} $$

La suma es:

$$ A+B= \begin{bmatrix} 1+5&2+6\\ 3+7&4+8 \end{bmatrix} $$

Por lo tanto:

$$ A+B= \begin{bmatrix} 6&8\\ 10&12 \end{bmatrix} $$

En Python:

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

C = A + B

print("Matriz A:")
print(A)

print("\nMatriz B:")
print(B)

print("\nA + B:")
print(C)
```

¿Qué pasa si las dimensiones son diferentes?

Por ejemplo:

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6, 7]
])

print(A.shape)
print(B.shape)
```

Aquí tenemos:

A → (2, 2)
B → (1, 3)

Estas estructuras no tienen la misma forma, por lo que no podemos realizar una suma matricial convencional entre ellas.

Resta de vectores

La resta de vectores se realiza componente por componente.

Si tenemos:

$$ A = \begin{bmatrix} a_1\\ a_2\\ a_3 \end{bmatrix} \quad B = \begin{bmatrix} b_1\\ b_2\\ b_3 \end{bmatrix} $$

entonces:

$$ A-B = \begin{bmatrix} a_1-b_1\\ a_2-b_2\\ a_3-b_3 \end{bmatrix} $$
Ejemplo 1: resta de dos vectores

Tenemos:

$$ A = \begin{bmatrix} 8\\ 6\\ 10 \end{bmatrix} \quad B = \begin{bmatrix} 3\\ 2\\ 4 \end{bmatrix} $$

Restamos cada componente:

$$ A-B = \begin{bmatrix} 8-3\\ 6-2\\ 10-4 \end{bmatrix} = \begin{bmatrix} 5\\ 4\\ 6 \end{bmatrix} $$


```python
import numpy as np

A = np.array([
    [8],
    [6],
    [10]
])

B = np.array([
    [3],
    [2],
    [4]
])

C = A - B

print("Vector A:")
print(A)

print("\nVector B:")
print(B)

print("\nA - B:")
print(C)
```

vectores horizontales

También podemos representar los vectores como una matriz de una fila:

$$ A = \begin{bmatrix} 10 & 8 & 6 \end{bmatrix} $$ $$ B = \begin{bmatrix} 2 & 3 & 4 \end{bmatrix} $$

Entonces:

$$ A-B = \begin{bmatrix} 10-2 & 8-3 & 6-4 \end{bmatrix} $$ $$ A-B = \begin{bmatrix} 8 & 5 & 2 \end{bmatrix} $$


```python
import numpy as np

A = np.array([[10, 8, 6]])
B = np.array([[2, 3, 4]])

C = A - B

print("A:")
print(A)

print("\nB:")
print(B)

print("\nA - B:")
print(C)
```

resta de matrices

La misma idea se aplica cuando tenemos matrices.

Tenemos:

$$ A = \begin{bmatrix} 10 & 8\\ 6 & 4 \end{bmatrix} $$

y

$$ B = \begin{bmatrix} 3 & 2\\ 1 & 2 \end{bmatrix} $$

Restamos cada posición:

$$ A-B = \begin{bmatrix} 10-3 & 8-2\\ 6-1 & 4-2 \end{bmatrix} $$

Por lo tanto:

$$ A-B = \begin{bmatrix} 7 & 6\\ 5 & 2 \end{bmatrix} $$


```python
import numpy as np

A = np.array([
    [10, 8],
    [6, 4]
])

B = np.array([
    [3, 2],
    [1, 2]
])

C = A - B

print("Matriz A:")
print(A)

print("\nMatriz B:")
print(B)

print("\nA - B:")
print(C)
```

Vectores transpuestos
1. ¿Qué es la transpuesta de un vector?

La transpuesta de un vector consiste en cambiar su orientación:

un vector columna se convierte en vector fila;
un vector fila se convierte en vector columna.

Se representa utilizando una T mayúscula:

$$ A^T $$

que se lee:

"A transpuesta".

Transpuesta

La transpuesta intercambia filas por columnas.

$$ A= \begin{bmatrix} 1&2&3\\ 4&5&6 \end{bmatrix} $$

Su transpuesta es:

$$ A^T= \begin{bmatrix} 1&4\\ 2&5\\ 3&6 \end{bmatrix} $$

Python:

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(A.T)
```

Transpuesta del vector columna

Si tenemos:

$$ A= \begin{bmatrix} 2\\ 4\\ 6 \end{bmatrix} $$

su transpuesta es:

$$ A^T= \begin{bmatrix} 2&4&6 \end{bmatrix} $$

Ahora tenemos:

1 fila
3 columnas

Por lo tanto:

A
T
→(1,3)

```python
import numpy as np

A = np.array([
    [2],
    [4],
    [6]
])

AT = A.T

print("Vector A:")
print(A)

print("\nTranspuesta de A:")
print(AT)

print("\nDimensiones de A:", A.shape)
print("Dimensiones de A.T:", AT.shape)
```

Concepto clave para IA
$$ \boxed{\text{La transpuesta cambia la orientación de un vector para facilitar operaciones de álgebra lineal.}} $$
