```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Funcion original
def f(x1, x2):
    return x1*2 + x2*2

# 2. Gradiente
def gradiente(x1, x2):
    return 2*x1, 2*x2

# 3. Parametros
beta = 0.45
x1 = 5.0
x2 = 5.0
tolerancia = 0.00000001

iteracion = 0

# Guardar los puntos para la grafica
historial_x1 = [x1]
historial_x2 = [x2]
historial_f = [f(x1, x2)]

# 4. Ejecutar Gradiente Descendente
print("GRADIENTE DESCENDENTE")
print("Funcion: f(x1,x2) = x1^2 + x2^2")
print("-----------------------------------")

while True:

    resultado = f(x1, x2)

    print("Iteracion:", iteracion)
    print("X1:", x1)
    print("X2:", x2)
    print("f(X1,X2):", resultado)
    print("-----------------------------------")

    if resultado < tolerancia:
        break

    # Calcular gradiente
    dx1, dx2 = gradiente(x1, x2)

    # Actualizar variables
    x1 = x1 - beta * dx1
    x2 = x2 - beta * dx2

    iteracion += 1

    # Guardar nuevos puntos
    historial_x1.append(x1)
    historial_x2.append(x2)
    historial_f.append(f(x1, x2))

print("\nMINIMO ENCONTRADO")
print("X1 =", x1)
print("X2 =", x2)
print("Valor minimo =", f(x1, x2))


# 5. Crear superficie 3D
X1, X2 = np.meshgrid(
    np.linspace(-1, 6, 100),
    np.linspace(-1, 6, 100)
)

Z = f(X1, X2)

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Dibujar superficie
ax.plot_surface(
    X1, X2, Z,
    cmap='viridis',
    alpha=0.7
)

# Dibujar trayectoria del algoritmo
ax.plot(
    historial_x1,
    historial_x2,
    historial_f,
    color='red',
    marker='o',
    markersize=6,
    linewidth=2,
    label='Trayectoria GD'
)

# Punto inicial
ax.scatter(
    5, 5, 50,
    color='blue',
    s=100,
    label='Punto inicial (5,5)'
)

# Punto minimo
ax.scatter(
    0, 0, 0,
    color='red',
    s=150,
    marker='*',
    label='Minimo (0,0,0)'
)

# Etiquetas
ax.set_xlabel('X1')
ax.set_ylabel('X2')
ax.set_zlabel('f(X1,X2)')

ax.set_title('Gradiente Descendente - Funcion 3D')

ax.legend()
plt.tight_layout()
plt.show()
```
