```python
# Definir la función compuesta

def funcion_interior(x):
    return x**2 + 1

def funcion_exterior(u):
    return u**3

# Valor de entrada
x = 2

# Evaluar primero la función interior
u = funcion_interior(x)

# Evaluar después la función exterior
y = funcion_exterior(u)

print("Valor de x:", x)
print("Resultado de la función interior:", u)
print("Resultado de la función exterior:", y)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Función original
def f(x):
    return (x**2 + 1)**3

# Derivada mediante la regla de la cadena
def df(x):
    return 6*x*(x**2 + 1)**2

# Punto de tangencia
a = 1
b = f(a)

# Pendiente de la tangente
m = df(a)

# Eje horizontal
x = np.linspace(-2, 2, 500)

# Función y recta tangente
y = f(x)
tangente = b + m*(x - a)

plt.figure(figsize=(10, 6))

plt.plot(x, y, label="f(x) = (x² + 1)³")
plt.plot(x, tangente, "--", label="Recta tangente en x = 1")
plt.scatter(a, b, color="red", s=80, label="Punto (1, 8)")

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Regla de la cadena: interpretación geométrica")
plt.grid(True)
plt.legend()
plt.ylim(-10, 150)

plt.show()

print("Punto de tangencia:", (a, b))
print("Pendiente:", m)
print("Ecuación de la tangente: y = 24x - 16")
```

```python
import sympy as sp

# Variable simbólica
x = sp.symbols('x')

# Lista de funciones
funciones = [
    (x**2 + 3)**5,
    (4*x**3 - 2*x + 1)**4,
    (3*x + 2)**7,
    (x**2 - 5*x + 4)**3
]

for f in funciones:
    derivada = sp.diff(f, x)

    print("Función original:")
    sp.pprint(f)

    print("Derivada:")
    sp.pprint(sp.factor(derivada))

    print("-" * 50)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Función original
def f(x):
    return (x**2 + 1)**3

# Derivada
def df(x):
    return 6*x*(x**2 + 1)**2

# Valores de x
x = np.linspace(-2, 2, 500)

# Calcular valores
y = f(x)
dy = df(x)

# Crear dos gráficas
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(x, y, label="f(x) = (x² + 1)³")
axes[0].scatter(0, f(0), color="red", label="Mínimo (0, 1)")
axes[0].set_title("Función original")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].grid(True)
axes[0].legend()

axes[1].plot(x, dy, label="f'(x) = 6x(x² + 1)²")
axes[1].axhline(0, color="black", linewidth=0.8)
axes[1].axvline(0, color="black", linewidth=0.8)
axes[1].set_title("Primera derivada")
axes[1].set_xlabel("x")
axes[1].set_ylabel("f'(x)")
axes[1].grid(True)
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
import sympy as sp

t = sp.symbols('t')

# Posición
s = (t**2 + 1)**3

# Velocidad
v = sp.diff(s, t)

# Aceleración
a = sp.diff(v, t)

print("Posición:")
sp.pprint(s)

print("\nVelocidad:")
sp.pprint(sp.factor(v))

print("\nAceleración:")
sp.pprint(sp.factor(a))

# Evaluar en t = 2
tiempo = 2

print("\nEn t = 2:")
print("Posición:", s.subs(t, tiempo))
print("Velocidad:", v.subs(t, tiempo))
print("Aceleración:", a.subs(t, tiempo))
```

```python
import numpy as np

# Datos de entrada y parámetros
x = 2.0
y_real = 1.0
w = 0.5
b = 0.1

# Función sigmoide
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Paso hacia adelante
z = w * x + b
y_predicho = sigmoid(z)

# Función de pérdida
loss = (y_predicho - y_real)**2

# Derivadas de cada etapa
dL_dyp = 2 * (y_predicho - y_real)
dyp_dz = y_predicho * (1 - y_predicho)
dz_dw = x

# Regla de la cadena
dL_dw = dL_dyp * dyp_dz * dz_dw

print("Valor de z:", z)
print("Predicción:", y_predicho)
print("Error cuadrático:", loss)
print("Derivada del error respecto al peso:", dL_dw)
```

```python
def f(x):
    return (x**2 + 1)**3

def df_exacta(x):
    return 6*x*(x**2 + 1)**2

def df_numerica(x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2*h)

x = 1.0

print("Valor de x:", x)
print("Derivada exacta:", df_exacta(x))
print("Derivada numérica:", df_numerica(x))
print("Diferencia:", abs(df_exacta(x) - df_numerica(x)))
```

# RESUMEN: REGLA DE LA CADENA

## Fórmula general

d/dx [f(g(x))] = f'(g(x)) · g'(x)

## Potencias

d/dx [u(x)]^n = n[u(x)]^(n-1) · u'(x)

## Seno

d/dx [sin(u)] = cos(u) · u'

## Coseno

d/dx [cos(u)] = -sin(u) · u'

## Tangente

d/dx [tan(u)] = sec²(u) · u'

## Exponencial

d/dx [e^u] = e^u · u'

## Logaritmo natural

d/dx [ln(u)] = u'/u

## Composición de tres funciones

Si y = f(g(h(x))), entonces:

y' = f'(g(h(x))) · g'(h(x)) · h'(x)
