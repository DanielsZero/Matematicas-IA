```python
import numpy as np
import matplotlib.pyplot as plt

# Definir los vectores
a = np.array([3, 2])
b = np.array([1, 4])

# Crear la gráfica
plt.figure(figsize=(8, 6))

plt.quiver(
    0, 0, a[0], a[1],
    angles="xy", scale_units="xy", scale=1,
    color="blue", label="Vector a"
)

plt.quiver(
    0, 0, b[0], b[1],
    angles="xy", scale_units="xy", scale=1,
    color="red", label="Vector b"
)

plt.xlim(-1, 5)
plt.ylim(-1, 6)

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("Componente X")
plt.ylabel("Componente Y")
plt.title("Representación de dos vectores")
plt.grid(True)
plt.legend()
plt.gca().set_aspect("equal", adjustable="box")

plt.show()
```

```python
# Componentes de los vectores
a = [2, 3]
b = [4, 1]

# Multiplicar componentes correspondientes
producto_1 = a[0] * b[0]
producto_2 = a[1] * b[1]

# Sumar los resultados
producto_punto = producto_1 + producto_2

print("Vector a:", a)
print("Vector b:", b)

print("Primer producto:", producto_1)
print("Segundo producto:", producto_2)

print("Producto punto:", producto_punto)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Vector fijo
a = np.array([3, 0])

# Diferentes ángulos
angulos = [45, 90, 135]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, angulo in zip(axes, angulos):

    rad = np.radians(angulo)

    # Vector b de magnitud 3
    b = np.array([
        3 * np.cos(rad),
        3 * np.sin(rad)
    ])

    # Producto punto
    producto = np.dot(a, b)

    ax.quiver(
        0, 0, a[0], a[1],
        angles="xy", scale_units="xy", scale=1,
        color="blue"
    )

    ax.quiver(
        0, 0, b[0], b[1],
        angles="xy", scale_units="xy", scale=1,
        color="red"
    )

    ax.set_xlim(-1, 4)
    ax.set_ylim(-1, 4)

    ax.axhline(0, color="black", linewidth=0.6)
    ax.axvline(0, color="black", linewidth=0.6)

    ax.set_title(
        f"Ángulo: {angulo}°\nProducto punto: {producto:.2f}"
    )

    ax.grid(True)
    ax.set_aspect("equal", adjustable="box")

plt.tight_layout()
plt.show()
```

```python
import numpy as np

a = np.array([1, 0])
b = np.array([1, 1])

# Producto punto
producto = np.dot(a, b)

# Magnitudes
magnitud_a = np.linalg.norm(a)
magnitud_b = np.linalg.norm(b)

# Verificar que ninguno sea el vector cero
if magnitud_a == 0 or magnitud_b == 0:
    print("No se puede calcular el ángulo con el vector cero.")
else:
    coseno = producto / (magnitud_a * magnitud_b)

    # Evitar pequeños errores numéricos
    coseno = np.clip(coseno, -1.0, 1.0)

    angulo = np.degrees(np.arccos(coseno))

    print("Producto punto:", producto)
    print("Magnitud de a:", magnitud_a)
    print("Magnitud de b:", magnitud_b)
    print("Ángulo:", angulo, "grados")
```

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([2, 3])
b = np.array([3, -2])

producto = np.dot(a, b)

print("Producto punto:", producto)

if np.isclose(producto, 0):
    print("Los vectores son ortogonales.")
else:
    print("Los vectores no son ortogonales.")

plt.figure(figsize=(7, 6))

plt.quiver(
    0, 0, a[0], a[1],
    angles="xy", scale_units="xy", scale=1,
    color="blue", label="Vector a"
)

plt.quiver(
    0, 0, b[0], b[1],
    angles="xy", scale_units="xy", scale=1,
    color="red", label="Vector b"
)

plt.xlim(-1, 4)
plt.ylim(-3, 4)

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.title("Vectores ortogonales")
plt.grid(True)
plt.legend()
plt.gca().set_aspect("equal", adjustable="box")

plt.show()
```

```python
import numpy as np

a = np.array([2, 3])
b = np.array([4, 1])
c = np.array([1, 2])

# Propiedad conmutativa
print("Conmutativa:")
print(np.dot(a, b), np.dot(b, a))

# Propiedad distributiva
izquierda = np.dot(a, b + c)
derecha = np.dot(a, b) + np.dot(a, c)

print("\nDistributiva:")
print("Lado izquierdo:", izquierda)
print("Lado derecho:", derecha)

# Producto de un vector consigo mismo
print("\nProducto de a consigo mismo:")
print(np.dot(a, a))

# Magnitud al cuadrado
print("Magnitud de a al cuadrado:")
print(np.linalg.norm(a)**2)
```

```python
import numpy as np

# Vectores como matrices columna
a = np.array([[2], [3]])
b = np.array([[4], [1]])

# Transpuesta de a multiplicada por b
resultado = a.T @ b

print("Vector a:")
print(a)

print("\nVector b:")
print(b)

print("\nTranspuesta de a:")
print(a.T)

print("\nProducto punto:")
print(resultado)

print("\nValor escalar:")
print(resultado.item())
```

```python
import numpy as np

# Fuerza en newtons
F = np.array([10, 0])

# Desplazamiento en metros
d = np.array([5, 0])

# Trabajo mecánico
trabajo = np.dot(F, d)

print("Fuerza:", F, "N")
print("Desplazamiento:", d, "m")
print("Trabajo realizado:", trabajo, "J")
```

```python
import numpy as np

# Vectores de ejemplo
texto_a = np.array([1, 2, 3])
texto_b = np.array([2, 4, 6])
texto_c = np.array([3, -1, 0])

def similitud_coseno(a, b):
    norma_a = np.linalg.norm(a)
    norma_b = np.linalg.norm(b)

    if norma_a == 0 or norma_b == 0:
        raise ValueError(
            "La similitud del coseno no está definida "
            "para vectores cero."
        )

    return np.dot(a, b) / (norma_a * norma_b)

print(
    "Similitud entre A y B:",
    similitud_coseno(texto_a, texto_b)
)

print(
    "Similitud entre A y C:",
    similitud_coseno(texto_a, texto_c)
)
```

```python
import numpy as np

print("CALCULADORA DE PRODUCTO PUNTO")

# Ingresar las componentes
a1 = float(input("Componente x del vector A: "))
a2 = float(input("Componente y del vector A: "))

b1 = float(input("Componente x del vector B: "))
b2 = float(input("Componente y del vector B: "))

# Crear vectores
a = np.array([a1, a2])
b = np.array([b1, b2])

# Producto punto
producto = np.dot(a, b)

print("\nVector A:", a)
print("Vector B:", b)
print("Producto punto:", producto)

# Clasificación
if np.isclose(np.linalg.norm(a), 0) or np.isclose(
    np.linalg.norm(b), 0
):
    print("El ángulo no se puede calcular con un vector cero.")

else:
    coseno = producto / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )

    coseno = np.clip(coseno, -1, 1)
    angulo = np.degrees(np.arccos(coseno))

    print("Ángulo entre los vectores:", angulo, "grados")

    if np.isclose(producto, 0):
        print("Los vectores son ortogonales.")
    elif producto > 0:
        print("El producto punto es positivo.")
    else:
        print("El producto punto es negativo.")
```
